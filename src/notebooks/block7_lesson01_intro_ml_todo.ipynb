## Что такое искусственный интеллект

Искусственный интеллект (AI) — это область информатики, которая занимается созданием систем, способных выполнять задачи, обычно требующие человеческого интеллекта.

Примеры:

- ChatGPT;
- голосовые помощники;
- рекомендации YouTube;
- рекомендации маркетплейсов;
- распознавание лиц;
- беспилотные автомобили.

---

## Что такое машинное обучение

Machine Learning (ML) — раздел искусственного интеллекта.

Вместо написания большого количества правил мы показываем компьютеру данные.

Компьютер сам ищет закономерности.

Пример:

Есть данные квартир:

- площадь;
- количество комнат;
- стоимость.

После обучения модель сможет предсказывать цену новой квартиры.

---

## Основные типы задач ML

### Регрессия

Предсказать число.

Примеры:

- стоимость квартиры;
- выручка магазина;
- температура воздуха.

### Классификация

Предсказать категорию.

Примеры:

- спам / не спам;
- купит / не купит;
- болен / здоров.

### Кластеризация

Найти группы без заранее известных ответов.

Примеры:

- сегментация клиентов;
- группировка товаров;
- анализ пользователей.

## Ячейка 1
Повторите задание самостоятельно.

````
import pandas as pd

df = pd.DataFrame({
    "area": [100, 200, 300, 400, 500, 600],
    "price": [1, 2, 3, 4, 5, 6]
})

display(df)

assert len(df) == 6
````

## Ячейка 2
Повторите задание самостоятельно.

````
X = df[["area"]]
y = df["price"]

print(X)
print(y)

assert X.shape == (6, 1)
````

## Ячейка 3
Повторите задание самостоятельно.

````
import matplotlib.pyplot as plt

plt.scatter(df["area"], df["price"])
plt.xlabel("Площадь")
plt.ylabel("Цена")
````

## Ячейка 4
Повторите задание самостоятельно.

````
model_name = "Linear Regression"
print(model_name)

assert model_name == "Linear Regression"
````

## Ячейка 5
Повторите задание самостоятельно.

````
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X, y)

assert model is not None
````

## Ячейка 6
Повторите задание самостоятельно.

````
import warnings
warnings.filterwarnings("ignore", category=UserWarning)

prediction = model.predict([[80]])

print(prediction)

assert prediction[0] > 0
````

## Ячейка 7
Повторите задание самостоятельно.

````
score = model.score(X, y)

print(score)

assert score > 0.9
````

## Ячейка 8
Повторите задание самостоятельно.

````
print(model.coef_)

preds = model.predict([[145], [155], [165]])

print(preds)

assert len(preds) == 3
````
