## Почему это важно

Большая часть времени в ML-проекте уходит не на обучение модели, а на подготовку данных.

Типичный ML-проект:

1. Сбор данных.
2. Очистка данных.
3. Подготовка признаков.
4. Разделение данных.
5. Обучение модели.
6. Оценка качества.

## Новые понятия

### Признаки (Features)

Информация, по которой модель принимает решение.

Пример квартиры:

- площадь;
- количество комнат;
- этаж.

### Целевая переменная (Target)

То, что модель должна предсказать.

Например:

- стоимость квартиры.

### Train и Test

Train — данные для обучения.

Test — данные для проверки.

Если тестировать модель на тех же данных, на которых она обучалась, результат будет обманчивым.

## Переобучение

Переобучение возникает, когда модель слишком хорошо запоминает обучающие данные и плохо работает на новых данных.

## Масштабирование

Некоторые алгоритмы работают лучше, если признаки приведены к одному масштабу.

Для этого используется StandardScaler.

## Связь с дипломом

Практически любой дипломный ML-проект будет использовать:

- train_test_split;
- масштабирование;
- оценку качества модели;
- контроль переобучения.

## Ячейка 1. Загрузка датасета недвижимости

### Алгоритм
Используем реальный датасет California Housing.

```
import pandas as pd

url = "../../data/CompleteDietDiary.csv"
df = pd.read_csv(url)

print(df.shape)

assert len(df) > 1000
```

## Ячейка 2. Изучение данных

### Алгоритм
Смотрим первые строки и колонки.

```
display(df.head())
print(df.columns.tolist())

assert "weight" in df.columns
```

## Ячейка 3. Выбор признаков и цели

### Алгоритм
Выделяем X и y.

```
feature_cols = ["protein_g", "fat_g", "carbohydrates_g"]
target_col = "weight"

X = df[feature_cols]
y = df[target_col]

print(X.shape)
print(y.shape)

assert len(X) == len(y)
```

## Ячейка 4. Обработка пропусков

### Алгоритм
Удаляем строки с пропусками.

```
data = df[[*feature_cols, target_col]].dropna()
print(data.shape)

assert len(data) > 1000
```

## Ячейка 5. Train/Test Split

### Алгоритм
Разделяем данные на обучение и тест.

```
from sklearn.model_selection import train_test_split
X = data[feature_cols]
y = data[target_col]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=621)

print(X_train.shape)
print(X_test.shape)

assert len(X_train) > len(X_test)
```

## Ячейка 6. Масштабирование признаков

### Алгоритм
Используем StandardScaler.

```
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(X_train_scaled[:3])

assert X_train_scaled.shape[0] == len(X_train)
```

## Ячейка 7. Первая модель

### Алгоритм
Обучаем линейную регрессию.

```
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train_scaled, y_train)

assert model is not None
```

## Ячейка 8. Оценка качества

### Алгоритм
Считаем R² на train и test.

```
train_score = model.score(X_train_scaled, y_train)
test_score = model.score(X_test_scaled, y_test)

print("Train:", train_score)
print("Test:", test_score)

assert train_score > 0
```

## Ячейка 9. Поиск переобучения

### Алгоритм
Сравниваем train и test.

```
difference = train_score - test_score

print("Разница:", difference)

assert difference < 0.5
```

## Ячейка 10. Мини-проект

### Алгоритм
Формируем итоговый отчет.

```
report = {
    "rows": len(data),
    "features": len(X.columns),
    "train_score": round(train_score,3),
    "test_score": round(test_score,3)
}

print(report)

assert report["features"] == 3
```
